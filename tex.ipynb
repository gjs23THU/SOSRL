## `State.alocate_op2sys` sequence check

This cell checks one operation assignment at a time: expected start/finish time, assignment table, system available time, task progress, and makespan.

```python
import importlib
import random
import traceback

import numpy as np

import syn
import env

random.seed(42)
np.random.seed(42)
importlib.reload(syn)
importlib.reload(env)

arch = syn.random_select_sos(12)
mission = syn.build_mission_from_config(syn.CONFIG)
mission
```

```python

real_release = np.array(
    [[op.release_time for op in task.operations] for task in mission],
    dtype=np.float32,
)
print("Current big-M:", state.M)
print("\nRelease-time initialization check")
print("state.op_release_time[:2]:")
print(state.op_release_time[:2])
print("mission release_time[:2]:")
print(real_release[:2])
print("state.op_release_time has nan:", bool(np.isnan(state.op_release_time).any()))
```

```python

if np.isnan(state.op_release_time).any():
    print("NOTE: this test repairs state.op_release_time for this notebook run only.")
    print("      In env.py, State.__init__ should use op.release_time, not op.start_time.")
    state.op_release_time = real_release.copy()


def find_next_assign_candidate(state, prefer_time_advance=True):
    candidates = []
    for task_idx in range(mission_env.T):
        op_idx = int(state.task_op_idx[task_idx])
        if op_idx >= mission_env.O:
            continue

        op = state.mission[task_idx].operations[op_idx]
        for sys_idx in range(mission_env.N):
            sys = env.FULL_SOS[sys_idx]
            if state.sys_sel_idx[sys_idx] != 1:
                continue
            if state.sys_work_idx[sys_idx] == 1:
                continue
            if sys.func_type != op.func_type:
                continue

            expected_start = max(
                float(state.sys_availble_time[sys_idx]),
                float(state.op_release_time[task_idx, op_idx]),
            )
            expected_finish = expected_start + float(op.duration)
            if expected_finish > sys.available_until:
                continue

            expected_current_time = max(float(state.current_time), expected_start)
            candidates.append((
                expected_current_time,
                expected_start,
                task_idx,
                op_idx,
                sys_idx,
                expected_finish,
            ))

    if not candidates:
        return None

    if prefer_time_advance and state.current_time == 0:
        advanced = [c for c in candidates if c[0] > state.current_time]
        if advanced:
            candidates = advanced

    candidates.sort()
    _, expected_start, task_idx, op_idx, sys_idx, expected_finish = candidates[0]
    return task_idx, op_idx, sys_idx, expected_start, expected_finish

```

```python
state.reset_state()
print("\nSequential allocation check")
for step in range(8):
    candidate = find_next_assign_candidate(state)
    if candidate is None:
        print(f"step {step}: no valid idle-system assignment candidate")
        break

    task_idx, op_idx, sys_idx, expected_start, expected_finish = candidate
    op = state.mission[task_idx].operations[op_idx]
    old_time = float(state.current_time)
    old_makespan = float(state.current_makespan)
    old_sys_available = float(state.sys_availble_time[sys_idx])
    expected_current_time = max(old_time, expected_start)
    expected_makespan = max(old_makespan, expected_finish)

    print("\n" + "-" * 72)
    print(f"step {step}: current time={old_time}, allocate task={task_idx}, op={op_idx}, sys={sys_idx}")
    print("sys:", {"func": env.FULL_SOS[sys_idx].func_type, "available_from": env.FULL_SOS[sys_idx].available_from, "available_until": env.FULL_SOS[sys_idx].available_until})
    print("op:", {"func": op.func_type, "duration": op.duration, "release": op.release_time})
    print("before:", {
        "current_time": old_time,
        "current_makespan": old_makespan,
        "sys_available": old_sys_available,
        "sys_work": int(state.sys_work_idx[sys_idx]),
        "task_op_idx": int(state.task_op_idx[task_idx]),
    })
    print("expected:", {
        "start": expected_start,
        "finish": expected_finish,
        "current_time": expected_current_time,
        "current_makespan": expected_makespan,
    })

    try:
        state.alocate_op2sys(task_idx, op_idx, sys_idx)
    except Exception as exc:
        print("alocate_op2sys raised:", type(exc).__name__, exc)
        traceback.print_exc(limit=4)
        break

    actual = {
        "assigned_sys": int(state.op_assign_sys[task_idx, op_idx]),
        "start": float(state.op_start_time[task_idx, op_idx]),
        "finish_recorded": float(state.op_finish_time[task_idx, op_idx]),
        "sys_available": float(state.sys_availble_time[sys_idx]),
        "current_time": float(state.current_time),
        "current_makespan": float(state.current_makespan),
        "sys_work": int(state.sys_work_idx[sys_idx]),
        "task_op_idx": int(state.task_op_idx[task_idx]),
    }
    print("actual:", actual)

    checks = {
        "assigned_sys": actual["assigned_sys"] == sys_idx,
        "start_time": np.isclose(actual["start"], expected_start),
        "finish_time_recorded": np.isclose(actual["finish_recorded"], expected_finish),
        "sys_available": np.isclose(actual["sys_available"], expected_finish),
        "current_time": np.isclose(actual["current_time"], expected_current_time),
        "current_makespan": np.isclose(actual["current_makespan"], expected_makespan),
        "task_op_idx": actual["task_op_idx"] == op_idx + 1,
    }
    print("checks:", checks)

print("\nFinal summary")
print("current_time:", state.current_time)
print("current_makespan:", state.current_makespan)
print("task_op_idx[:8]:", state.task_op_idx[:8].tolist())
print("op_assign_sys[:4]:")
print(state.op_assign_sys[:4])
print("op_start_time[:4]:")
print(state.op_start_time[:4])
print("op_finish_time[:4]:")
print(state.op_finish_time[:4])
```

## Assign mask check for definition 2

This cell recomputes the expected assign mask independently and compares it with `mask_invalid_assign()`.

```python
import importlib
import random

import numpy as np

import syn
import env

random.seed(42)
np.random.seed(42)
importlib.reload(syn)
importlib.reload(env)


def expected_assign_mask(mission_env):
    expected = np.zeros((mission_env.T, mission_env.O, mission_env.N), dtype=np.float32)
    state = mission_env.state

    for task_idx in range(mission_env.T):
        op_idx = int(state.task_op_idx[task_idx])
        if op_idx >= mission_env.O:
            continue

        op = state.mission[task_idx].operations[op_idx]
        for sys_idx in range(mission_env.N):
            sys = env.FULL_SOS[sys_idx]
            if state.sys_sel_idx[sys_idx] != 1:
                continue
            if sys.func_type != op.func_type:
                continue

            start_time = max(
                float(state.current_time),
                float(state.sys_availble_time[sys_idx]),
                float(state.op_release_time[task_idx, op_idx]),
            )
            finish_time = start_time + float(op.duration)
            if finish_time <= float(sys.available_until):
                expected[task_idx, op_idx, sys_idx] = 1

    return expected


def print_valid_examples(mission_env, assign_mask, limit=5):
    state = mission_env.state
    valid = np.argwhere(assign_mask == 1)
    print("valid assign count:", len(valid))

    for task_idx, op_idx, sys_idx in valid[:limit]:
        op = state.mission[int(task_idx)].operations[int(op_idx)]
        sys = env.FULL_SOS[int(sys_idx)]
        start_time = max(
            float(state.current_time),
            float(state.sys_availble_time[sys_idx]),
            float(state.op_release_time[task_idx, op_idx]),
        )
        finish_time = start_time + float(op.duration)
        print({
            "task": int(task_idx),
            "op": int(op_idx),
            "sys": int(sys_idx),
            "func": op.func_type,
            "current_time": float(state.current_time),
            "sys_available": float(state.sys_availble_time[sys_idx]),
            "release": float(state.op_release_time[task_idx, op_idx]),
            "start": start_time,
            "finish": finish_time,
            "sys_until": float(sys.available_until),
        })


def check_assign_mask(mission_env, label):
    actual = mission_env.mask_invalid_assign()
    expected = expected_assign_mask(mission_env)
    mismatch = np.argwhere(actual != expected)

    print("\n" + "=" * 80)
    print(label)
    print("actual shape:", actual.shape)
    print("expected shape:", expected.shape)
    print("actual valid:", int(actual.sum()))
    print("expected valid:", int(expected.sum()))
    print("mismatch count:", len(mismatch))

    if len(mismatch) > 0:
        print("first mismatches:", mismatch[:10].tolist())
        for task_idx, op_idx, sys_idx in mismatch[:5]:
            print({
                "task": int(task_idx),
                "op": int(op_idx),
                "sys": int(sys_idx),
                "actual": float(actual[task_idx, op_idx, sys_idx]),
                "expected": float(expected[task_idx, op_idx, sys_idx]),
            })
    else:
        print("mask check: PASS")

    print_valid_examples(mission_env, actual)


arch = syn.random_select_sos(12)
mission = syn.build_mission_from_config(syn.CONFIG)
mission_env = env.MissionEnv(arch=arch, mission=mission)
check_assign_mask(mission_env, "Scenario 1: initial state")

# Scenario 2 checks that updated op_release_time participates in the mask for a non-first op.
mission_env2 = env.MissionEnv(arch=arch, mission=mission)
state = mission_env2.state
state.current_time = 50.0
state.task_op_idx[0] = 1
state.op_assign_sys[0, 0] = 0
state.op_release_time[0, 1] = 120.0
check_assign_mask(mission_env2, "Scenario 2: task 0 moves to op 1 with op_release_time = 120")

full_mask = mission_env.mask_invalid_actions()
assign_size = mission_env.T * mission_env.O * mission_env.N
print("\nFull action mask")
print("action_space.n:", mission_env.action_space.n)
print("full mask shape:", full_mask.shape)
print("full mask matches action_space:", full_mask.shape == (mission_env.action_space.n,))
print("assign/select/deselect valid:", int(full_mask[:assign_size].sum()), int(full_mask[assign_size:assign_size + mission_env.N].sum()), int(full_mask[assign_size + mission_env.N:].sum()))
```

## Earliest-start assignment sequence check

This test does not change `env.py`. It only chooses a reasonable sequence from the current action mask: pick the valid assign action with the earliest feasible start time, then earliest finish time.

```python
import importlib
import random

import numpy as np

import syn
import env

random.seed(1)
np.random.seed(1)
importlib.reload(syn)
importlib.reload(env)


def choose_earliest_assign(mission_env):
    assign_mask = mission_env.mask_invalid_assign()
    candidates = np.argwhere(assign_mask == 1)
    if len(candidates) == 0:
        return None, None

    best = None
    best_key = None
    for task_idx, op_idx, sys_idx in candidates:
        task_idx = int(task_idx)
        op_idx = int(op_idx)
        sys_idx = int(sys_idx)
        op = mission_env.state.mission[task_idx].operations[op_idx]

        start_time = max(
            float(mission_env.state.current_time),
            float(mission_env.state.sys_availble_time[sys_idx]),
            float(mission_env.state.op_release_time[task_idx, op_idx]),
        )
        finish_time = start_time + float(op.duration)
        key = (start_time, finish_time, float(mission_env.state.sys_busy_time[sys_idx]), task_idx, sys_idx)

        if best_key is None or key < best_key:
            best_key = key
            best = (task_idx, op_idx, sys_idx, start_time, finish_time)

    action = mission_env.encode_action(
        "assign_task",
        task_idx=best[0],
        op_idx=best[1],
        sys_idx=best[2],
    )
    return int(action), best


def run_earliest_sequence(arch, mission, label, max_steps=300, verbose_first=12):
    mission_env = env.MissionEnv(arch=arch, mission=mission)
    obs, _ = mission_env.reset()
    history = []

    for step in range(max_steps):
        action, decoded = choose_earliest_assign(mission_env)
        if action is None:
            print(f"{label}: no valid assign at step {step}")
            break

        obs, reward, terminated, truncated, info = mission_env.step(action)
        history.append(decoded)

        if step < verbose_first or terminated or truncated:
            print({
                "step": step,
                "chosen": decoded,
                "current_time": float(mission_env.state.current_time),
                "makespan": float(mission_env.state.current_makespan),
                "done_ops": int(mission_env.state.task_op_idx.sum()),
                "terminated": bool(terminated),
                "truncated": bool(truncated),
            })

        if terminated or truncated:
            break

    print("\nsummary:", {
        "label": label,
        "done_ops": int(mission_env.state.task_op_idx.sum()),
        "total_ops": mission_env.T * mission_env.O,
        "current_time": float(mission_env.state.current_time),
        "makespan": float(mission_env.state.current_makespan),
        "selected_systems": int(mission_env.state.sys_sel_idx.sum()),
        "obs_in_space": bool(mission_env.observation_space.contains(obs)),
    })
    return mission_env, history


mission = syn.build_mission_from_config(syn.CONFIG)
random_arch = syn.random_select_sos(12)

random_arch_env, random_arch_history = run_earliest_sequence(random_arch, mission, "random 12-system arch")
full_arch_env, full_arch_history = run_earliest_sequence(env.FULL_SOS, mission, "full SOS arch")
```

## Fixed-rule rollout check

Use one fixed paper rule action, convert it through `Rule.to_env_action()`, and let the environment execute the resulting `assign_task` action until termination or no feasible assign remains.


```python
import importlib
import random

import numpy as np

import syn
import env
import rule

random.seed(1)
np.random.seed(1)
importlib.reload(syn)
importlib.reload(env)
importlib.reload(rule)

fixed_rule_action = 4 # 0-4, corresponding to the five paper rules


def clean_decoded_action(decoded):
    cleaned = dict(decoded)
    for key in ("task_idx", "op_idx", "sys_idx"):
        if key in cleaned:
            cleaned[key] = int(cleaned[key])
    return cleaned


def run_fixed_rule_schedule(arch, mission, fixed_rule_action=0, max_steps=300, verbose_first=30):
    mission_env = env.MissionEnv(arch=arch, mission=mission)
    obs, _ = mission_env.reset()
    rule_policy = rule.Rule(mission_env)
    history = []

    for step in range(max_steps):
        try:
            action = rule_policy.to_env_action(fixed_rule_action)
        except ValueError as exc:
            print(f"no env action at step {step}: {exc}")
            break

        decoded = clean_decoded_action(mission_env.decode_action(action))
        obs, reward, terminated, truncated, info = mission_env.step(action)
        history.append(decoded)

        if step < verbose_first or terminated or truncated:
            print({
                "step": step,
                "rule": fixed_rule_action,
                "env_action": int(action),
                "decoded": decoded,
                "reward": float(reward),
                "current_time": float(mission_env.state.current_time),
                "current_action_space": int(mission_env.mask_invalid_actions().sum()),
                "makespan": float(mission_env.state.current_makespan),
                "done_ops": int(mission_env.state.task_op_idx.sum()),
                "terminated": bool(terminated),
                "truncated": bool(truncated),
            })

        if terminated or truncated:
            break

    print("\nsummary:", {
        "fixed_rule_action": fixed_rule_action,
        "steps": len(history),
        "done_ops": int(mission_env.state.task_op_idx.sum()),
        "total_ops": mission_env.T * mission_env.O,
        "action_space": int(mission_env.mask_invalid_actions().sum()),
        "current_time": float(mission_env.state.current_time),
        "makespan": float(mission_env.state.current_makespan),
        "selected_systems": int(mission_env.state.sys_sel_idx.sum()),
        "obs_in_space": bool(mission_env.observation_space.contains(obs)),
    })
    return mission_env, history


mission = syn.build_mission_from_config(syn.CONFIG)
fixed_rule_env, fixed_rule_history = run_fixed_rule_schedule(
    arch=env.FULL_SOS,
    mission=mission,
    fixed_rule_action=fixed_rule_action,
)
```
